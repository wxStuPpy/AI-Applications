```python
from langchain_core.output_parsers import StrOutputParser
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.runnables import RunnableLambda
from langchain_openai import ChatOpenAI
import os
import dotenv
dotenv.load_dotenv()

os.environ["OPENAI_API_KEY"] = os.getenv("OPENAI_API_KEY1")
os.environ["OPENAI_BASE_URL"] = os.getenv("OPENAI_BASE_URL")

model=ChatOpenAI(
    model='gpt-4o-mini',
    temperature=0.8,
    max_tokens=20
)
# 一个简单的打印函数，调试用
def debug_print(x):
    print("中间结果:", x)
    return x

# 子链1提示词
news_prompt = ChatPromptTemplate.from_messages([
    ("system", "你是一个知识渊博的计算机专家，请用中文回答"),
    ("human", "请简短介绍什么是{name}")
])
# 子链1解析器
news_parser = StrOutputParser()
# 子链1：生成新闻内容
news_chain = news_prompt | model | news_parser

# 子链2提示词
translate_prompt = ChatPromptTemplate.from_messages([
    ("system", "你是一个翻译助手，将用户输入内容翻译成英文"),
    ("human", "{content}")
])
# 子链2解析器
translate_parser = StrOutputParser()

# 子链2：翻译内容
translate_chain = translate_prompt | model | translate_parser
# 添加debug节点
debug_node = RunnableLambda(debug_print)
# 组合成一个复合 Chain，使用 lambda 函数将字符串转换为字典，并将结果作为参数传递给 translate_chain
full_chain = news_chain | debug_node | (lambda content: {"content": content}) | translate_chain

# 调用复合链
result = full_chain.invoke({"name": "langchain"})
print(result)
```
