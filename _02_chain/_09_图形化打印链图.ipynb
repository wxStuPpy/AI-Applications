```python
from langchain_core.output_parsers import StrOutputParser
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.runnables import RunnableParallel, RunnableLambda
from langchain_openai import ChatOpenAI
import os
import dotenv
dotenv.load_dotenv()

os.environ["OPENAI_API_KEY"] = os.getenv("OPENAI_API_KEY1")
os.environ["OPENAI_BASE_URL"] = os.getenv("OPENAI_BASE_URL")

model=ChatOpenAI(
    model='gpt-4o-mini',
    temperature=0.8,
    max_tokens=100
)


# 一个简单的打印函数，调试用
def debug_print(x):
    print("中间结果:", x)
    return x

# 子链1提示词
chinese_prompt = ChatPromptTemplate.from_messages([
    ("system", "你是一个知识渊博的计算机专家，请用中文回答"),
    ("human", "请简短介绍什么是{chinese_name}")
])
# 子链1解析器
chinese_parser = StrOutputParser()
# 子链1：生成新闻内容
chinese_chain = chinese_prompt | model | chinese_parser

# 子链2提示词
english_prompt = ChatPromptTemplate.from_messages([
    ("system", "你是一个知识渊博的计算机专家，请用英文回答"),
    ("human", "请简短介绍什么是{english_name}")
])
# 子链2解析器
english_parser = StrOutputParser()

# 子链2：翻译内容
english_chain = english_prompt | model | english_parser
# 添加debug节点
debug_node = RunnableLambda(debug_print)
# 创建并行链
parallel_chain = RunnableParallel({
    "chinese": chinese_chain | debug_node,
    "english": english_chain
})

# 调用复合链
result = parallel_chain.invoke({"chinese_name": "langchain", "english_name": "langchain"})
parallel_chain.get_graph().print_ascii()
print(result)
```
