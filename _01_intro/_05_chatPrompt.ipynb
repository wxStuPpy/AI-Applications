提示词模版之ChatPromptTemplate的使用

```python
from langchain_core.messages import HumanMessage, AIMessage
from langchain_core.prompts.chat import MessagesPlaceholder
#1实例化方式
#方式一:使用构造方法
from langchain_core.prompts import ChatPromptTemplate

chat_prompt_template = ChatPromptTemplate(messages=[
    ('system', '你是一个AI助手，你的名字是{name}'),
    ('human', '我的问题是{question}')
],
    #input_variables=['name', 'question']
)
```

```python
res=chat_prompt_template.invoke(input={'name':'kun','question':'1+1等于几'})
print(res)
print(type(res))
print(len(res.messages))
```

```python
#方式二：调用from_messages()
chat_prompt_template = ChatPromptTemplate.from_messages(
    messages=[
        ('system', '你是一个AI助手，你的名字是{name}'),
        ('human', '我的问题是{question}')
    ])

res=chat_prompt_template.invoke(input={'name':'kun','question':'1+1等于几'})
print(res)
print(type(res))
print(len(res.messages))
```

```python
#2.调用提示词模板的几种方式 invoke(),format(),format_messages(),format_prompt()
#方式一:invoke 传入字典 返回ChatPromptValue
chat_prompt_template = ChatPromptTemplate.from_messages(
    messages=[
        ('system', '你是一个AI助手，你的名字是{name}'),
        ('human', '我的问题是{question}')
    ])

res=chat_prompt_template.invoke(input={'name':'kun','question':'1+1等于几'})
print(res)
print(type(res))
print(len(res.messages))
```

```python
#方式二:format 传入变量值 返回字符串
res=chat_prompt_template.format(name='kun',question='1+1=?')
print(res)
print(type(res))
```

```python
#3.方式三:format_messages 传入变量值 返回消息构成的列表
res=chat_prompt_template.format_messages(name='kun',question='1+1=?')
print(res)
print(type(res))
```

```python
#4.方式四:format_prompt  传入变量值 返回ChatPromptValue 等同于invoke 传入参数不同而已
res=chat_prompt_template.format_prompt(name='kun',question='1+1=?')
print(res)
print(type(res))
#res=res.to_string()   将ChatPromptValue转为字符串
#将ChatPromptValue转为消息构成的列表
res=res.to_messages()
print(type(res))
```

```python
#4.结合llm
from langchain_openai import ChatOpenAI
from langchain_core.prompts import PromptTemplate
import os
import dotenv

dotenv.load_dotenv()

os.environ["OPENAI_API_KEY"] = os.getenv("OPENAI_API_KEY1")
os.environ["OPENAI_BASE_URL"] = os.getenv("OPENAI_BASE_URL")

model=ChatOpenAI(
    temperature=0.7,
    max_tokens=50
)

chat_prompt_template = ChatPromptTemplate.from_messages(
    messages=[
        ('system', '你是一个AI助手，你的名字是{name}'),
        ('human', '我的问题是{question}')
    ])
prompt_response=chat_prompt_template.invoke(input={'name':'kun','question':'1+1=?'})

res=model.invoke(prompt_response)
print(res.content)
print(type(res))
```

```python
#5.插入消息列表:MessagePlaceholder
chat_prompt_template = ChatPromptTemplate.from_messages(
    messages=[
        ('system', '你是一个AI助手，你的名字是{name}'),
        #('human', '我的问题是{question}')
        MessagesPlaceholder(variable_name='msgs')#占位 目前不知道是什么消息类型
    ])

res = chat_prompt_template.invoke(input={'name': 'kun',
                                         'msgs': [HumanMessage(content='我的问题是:1+1=?'),
                                                  AIMessage(content='1+1=2')]})
print(res)
```

```python
#6.存储对话历史记录
chat_prompt_template = ChatPromptTemplate.from_messages(
    messages=[
        ('system', '你是一个AI助手，你的名字是{name}'),
        ('human', '我的问题是{question}'),
        MessagesPlaceholder(variable_name='history')
    ])
prompt_res = chat_prompt_template.invoke(input={'name': 'kun',
                                         'history': [HumanMessage(content='我的问题是:1+1=?'),
                                                     AIMessage(content='1+1=2')], 'question': '我刚才问了什么问题?'})
res=model.invoke(prompt_res)
print(res.content)
```
