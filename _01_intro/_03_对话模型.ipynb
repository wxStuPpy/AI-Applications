```python
from langchain_openai import ChatOpenAI
import os
import dotenv

dotenv.load_dotenv()

os.environ["OPENAI_API_KEY"] = os.getenv("OPENAI_API_KEY1")
os.environ["OPENAI_BASE_URL"] = os.getenv("OPENAI_BASE_URL")
```

```python
llm=ChatOpenAI(
    temperature=0.8,
    model='gpt-4o-mini',
    max_tokens=100
)
```

invoke的输入有字符串和消息列表两种类型

```python
res=llm.invoke('肯德基有疯狂星期几')
print(res.content)
```

关于消息 有三种典型的消息 SystemMessage\HumanMessage\AIMessage

```python
from langchain_core.messages import SystemMessage,HumanMessage

system_msg=SystemMessage(content='你是一个数学专家')
human_msg=HumanMessage(content='帮我指定高三数学的复习计划')

msgs=[system_msg,human_msg]
print(msgs)
```

```python
#使用大模型 调用消息列表
res=llm.invoke(msgs)
print(res.content)
```

模型调用的方法 invoke/stream/batch/ainvoke/astream

```python
#1.invoke 阻塞式调用 一次性生成
```

```python
#2.流式调用
llm_stream=ChatOpenAI(
    temperature=0.8,
    model='gpt-4o-mini',
    streaming=True,
)

msg=[HumanMessage(content='你好，请介绍一下自己')]
for chunk in llm_stream.stream(msg):
    print(chunk.content,end='',flush=True)
```

```python
#3.batch批量调用
# 初始化模型（注意：streaming=True 适合流式输出，批量调用可设为 False 提高效率）
llm_batch = ChatOpenAI(
    temperature=0.8,
    model='gpt-4o-mini',
)

# 定义批量查询（多个问题）
batch_queries = [
    "什么是人工智能？用一句话概括",
    "推荐3个适合初学者的Python库",
    "解释一下什么是机器学习中的过拟合"
]

# 将查询转换为模型可接受的格式（HumanMessage 列表）
messages_list = [[HumanMessage(content=query)] for query in batch_queries]

# 批量调用模型（batch 方法支持批量处理）
results = llm_batch.batch(messages_list)
print(results)
```
