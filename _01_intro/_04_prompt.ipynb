```python
from langchain_openai import ChatOpenAI
from langchain_core.prompts import PromptTemplate
import os
import dotenv

dotenv.load_dotenv()

os.environ["OPENAI_API_KEY"] = os.getenv("OPENAI_API_KEY1")
os.environ["OPENAI_BASE_URL"] = os.getenv("OPENAI_BASE_URL")
```

```python
#1.创建PromptTemplate实例
#方式一 构造
#参数中必须要指明 input_variables template
prompt_template1=PromptTemplate(
    template='你是一个{role},你的名字是{name}',
    input_variables=['role','name']
)

print(prompt_template1)
```

```python
prompt=prompt_template1.format(role='数学家',name='人机')
print(prompt)
```

```python
#方式二 from_template 推荐
prompt_template2 = PromptTemplate.from_template(template='你是一个{role},你的名字是{name}')
print(prompt_template2)
```

```python
prompt=prompt_template2.format(role='数学家',name='人机')
print(prompt)
```

```python
#2.两种特殊结构的使用
#2.1.1构造时 部分提示词模板的使用(设置默认值)
template=PromptTemplate.from_template(
    template='你是一个{role},你的名字是{name}',
    partial_variables={'role':'坤'}
)

prompt=template.format(name='man ba')
print(prompt)
prompt=template.format(role='aaa',name='man ba')
print(prompt)
```

```python
#2.1.2 利用partial接口
template=PromptTemplate(
    template='你是一个{role},你的名字是{name}',
    input_variables=['role','name']
)
#返回一个新的模版 不改变原始Template
template=template.partial(role='bbb')
prompt=template.format(name='man ba')
print(prompt)
```
