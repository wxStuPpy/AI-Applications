```python
from langchain_core.tools import tool
from pydantic import BaseModel,Field

class FieldInfo(BaseModel):
    a:int=Field(description='第一个参数')
    b:int=Field(description='第二个参数')

@tool(return_direct=True,description='返回两数之和',args_schema=FieldInfo)
def add_num(a: int, b: int) -> int:
    """
    用于计算两个整数的和。
    """
    return a + b
```

```python
print(f'name={add_num.name}')
print(f'desc={add_num.description}')
print(f'args={add_num.args}')
print(f'returns_direct={add_num.return_direct}')
```

```python
#调用工具
res=add_num.invoke({'a':1, 'b':2})
print(res)
```
